```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler

TARGET   = 'stress_level'
FEATURES = [c for c in df.columns if c != TARGET]

X = df[FEATURES].values
y = df[TARGET].values

# ── Step 1: Carve out 10% test set first ─────────────────────────────
X_temp, X_test, y_temp, y_test = train_test_split(
    X, y,
    test_size=0.10,
    random_state=42,
    stratify=y          # preserves class proportions in every split
)

# ── Step 2: From remaining 90%, take ~22.2% as validation ────────────
# 20/90 = 0.222 → gives us exact 70:20:10 of original 1100
X_train, X_val, y_train, y_val = train_test_split(
    X_temp, y_temp,
    test_size=0.222,
    random_state=42,
    stratify=y_temp
)

# ── Verify split ──────────────────────────────────────────────────────
total = len(X)
print("="*55)
print("DATA SPLIT SUMMARY")
print("="*55)
print(f"Total samples    : {total}")
print(f"Train set        : {len(X_train):4d}  ({len(X_train)/total*100:.1f}%)")
print(f"Validation set   : {len(X_val):4d}  ({len(X_val)/total*100:.1f}%)")
print(f"Test set         : {len(X_test):4d}  ({len(X_test)/total*100:.1f}%)")

print(f"\nClass distribution per split:")
print(f"{'Class':<8} {'Train':>8} {'Val':>8} {'Test':>8}")
print("─"*35)
for cls in sorted(np.unique(y)):
    tr  = np.sum(y_train == cls)
    val = np.sum(y_val   == cls)
    te  = np.sum(y_test  == cls)
    print(f"  {cls}      {tr:>6}    {val:>6}    {te:>6}")
print("--"*35)
print("\n Split complete. Test set locked — will not be touched until final evaluation.")
print(f"\nFeatures : {len(FEATURES)}")
print(f"Feature names : {FEATURES}")
```

```python
from sklearn.preprocessing import MinMaxScaler

# ── Fit ONLY on training set ──────────────────────────────────────────
scaler = MinMaxScaler()
X_train_scaled = scaler.fit_transform(X_train)   # fit + transform
X_val_scaled   = scaler.transform(X_val)          # transform only
X_test_scaled  = scaler.transform(X_test)         # transform only

# ── Verify scaling ────────────────────────────────────────────────────
print("="*55)
print("SCALING VERIFICATION")
print("="*55)
print(f"{'Split':<18} {'Min':>8} {'Max':>8} {'Mean':>8} {'Std':>8}")
print("─"*50)
for name, arr in [('Train (scaled)', X_train_scaled),
                   ('Val   (scaled)', X_val_scaled),
                   ('Test  (scaled)', X_test_scaled)]:
    print(f"{name:<18} {arr.min():>8.4f} {arr.max():>8.4f} "
          f"{arr.mean():>8.4f} {arr.std():>8.4f}")

print(f"\nOriginal train range  : [{X_train.min():.2f}, {X_train.max():.2f}]")
print(f"Scaled train range    : [{X_train_scaled.min():.4f}, {X_train_scaled.max():.4f}]")

# ── Quick visual check ────────────────────────────────────────────────
fig, axes = plt.subplots(1, 2, figsize=(14, 4))

# Before scaling — first 5 features
axes[0].boxplot([X_train[:, i] for i in range(5)],
                labels=[FEATURES[i][:10] for i in range(5)],
                patch_artist=True)
axes[0].set_title('Before Scaling (first 5 features)', fontweight='bold')
axes[0].set_ylabel('Raw Value')
axes[0].tick_params(axis='x', rotation=30)
axes[0].grid(True, alpha=0.3)

# After scaling — first 5 features
bp = axes[1].boxplot([X_train_scaled[:, i] for i in range(5)],
                      labels=[FEATURES[i][:10] for i in range(5)],
                      patch_artist=True)
for patch in bp['boxes']:
    patch.set_facecolor('#4CAF50')
    patch.set_alpha(0.7)
axes[1].set_title('After MinMax Scaling (first 5 features)', fontweight='bold')
axes[1].set_ylabel('Scaled Value [0, 1]')
axes[1].set_ylim(-0.05, 1.05)
axes[1].axhline(y=0, color='red',   linestyle='--', linewidth=0.8, alpha=0.5)
axes[1].axhline(y=1, color='red',   linestyle='--', linewidth=0.8, alpha=0.5)
axes[1].tick_params(axis='x', rotation=30)
axes[1].grid(True, alpha=0.3)

plt.suptitle('MinMax Scaling Effect', fontsize=12, fontweight='bold')
plt.tight_layout()
plt.savefig('scaling_verification.png', dpi=150, bbox_inches='tight')
plt.show()

# ── Save scaler for later use ─────────────────────────────────────────
import joblib
joblib.dump(scaler, 'minmax_scaler.pkl')

print("\n✅ Scaling complete!")
print("   scaler saved → minmax_scaler.pkl")
print("\nNext: Step 5 — CTGAN Data Augmentation")
```

```python
import pandas as pd
import numpy as np
from sdv.single_table import CTGANSynthesizer
from sdv.metadata import SingleTableMetadata
from imblearn.over_sampling import SMOTE
from collections import Counter
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

print("="*55)
print("STEP 5 — CTGAN DATA AUGMENTATION")
print("="*55)

# ── Prepare training dataframe for CTGAN ─────────────────────────────
# CTGAN works on dataframes, use SCALED training data
train_df = pd.DataFrame(X_train_scaled, columns=FEATURES)
train_df['stress_level'] = y_train.astype(int)

print(f"Input to CTGAN:")
print(f"  Rows    : {len(train_df)}")
print(f"  Columns : {train_df.shape[1]}")
print(f"  Class distribution: {dict(Counter(y_train))}")

# ── Auto-detect metadata ──────────────────────────────────────────────
metadata = SingleTableMetadata()
metadata.detect_from_dataframe(train_df)

# stress_level is categorical (0,1,2) — set it explicitly
metadata.update_column(
    column_name='stress_level',
    sdtype='categorical'
)

print("\nMetadata detected ✅")
print("Training CTGAN — this will take 3–6 minutes on your GPU...")
print("(300 epochs, learning from the joint distribution of all 21 features)\n")

# ── Train CTGAN ───────────────────────────────────────────────────────
synthesizer = CTGANSynthesizer(
    metadata,
    epochs=500,
    batch_size=500,
    generator_dim=(256, 256, 128),
    discriminator_dim=(256, 256, 128),
    generator_lr=2e-4,
    discriminator_lr=2e-4,
    pac=10,
    verbose=True
)
synthesizer.fit(train_df)

print("\n✅ CTGAN training complete!")

# ── Generate synthetic samples ────────────────────────────────────────
# Generate 12000 synthetic + 770 original = ~12770 total
# SMOTE will balance to exactly ~15k with perfect class balance
N_SYNTHETIC = 12000
synthetic_df = synthesizer.sample(num_rows=N_SYNTHETIC)

# Clip to [0,1] range — CTGAN may occasionally go slightly outside
feat_cols = FEATURES
synthetic_df[feat_cols] = synthetic_df[feat_cols].clip(0.0, 1.0)

print(f"\nSynthetic samples generated : {len(synthetic_df)}")
print(f"Synthetic class distribution: {dict(synthetic_df['stress_level'].value_counts().sort_index())}")

# ── Combine original + synthetic training data ────────────────────────
X_syn  = synthetic_df[feat_cols].values
y_syn  = synthetic_df['stress_level'].values.astype(int)

X_combined = np.vstack([X_train_scaled, X_syn])
y_combined = np.concatenate([y_train, y_syn])

print(f"\nAfter combining original + synthetic:")
print(f"  Total rows          : {len(X_combined)}")
print(f"  Class distribution  : {dict(Counter(y_combined))}")

# ── SMOTE to perfectly balance classes ───────────────────────────────
print("\nApplying SMOTE to balance classes perfectly...")
smote = SMOTE(random_state=42, k_neighbors=5)
X_train_aug, y_train_aug = smote.fit_resample(X_combined, y_combined)

print(f"\nFinal augmented training set:")
print(f"  Total rows          : {len(X_train_aug)}")
print(f"  Class distribution  : {dict(Counter(y_train_aug))}")
print(f"  Original train size : {len(X_train_scaled)}")
print(f"  Growth factor       : {len(X_train_aug)/len(X_train_scaled):.1f}x")

# ── Validate synthetic data quality ──────────────────────────────────
print("\n── Statistical Fidelity Check ──────────────────────────")
print(f"{'Feature':<28} {'Orig Mean':>10} {'Syn Mean':>10} {'Diff':>8}")
print("─"*60)
for i, feat in enumerate(FEATURES):
    orig_mean = X_train_scaled[:, i].mean()
    syn_mean  = X_syn[:, i].mean()
    diff      = abs(orig_mean - syn_mean)
    flag      = " ⚠️" if diff > 0.1 else ""
    print(f"{feat:<28} {orig_mean:>10.4f} {syn_mean:>10.4f} {diff:>8.4f}{flag}")

# ── Visualization ─────────────────────────────────────────────────────
fig, axes = plt.subplots(2, 3, figsize=(18, 10))

# 1. Class distribution before augmentation
ax = axes[0, 0]
orig_counts = [np.sum(y_train == c) for c in [0, 1, 2]]
bars = ax.bar(['Low(0)', 'Mod(1)', 'High(2)'], orig_counts,
              color=['#4CAF50','#FF9800','#F44336'], edgecolor='black')
for bar, val in zip(bars, orig_counts):
    ax.text(bar.get_x()+bar.get_width()/2, bar.get_height()+3,
            str(val), ha='center', fontweight='bold')
ax.set_title('Training Set — BEFORE Augmentation', fontweight='bold')
ax.set_ylabel('Count'); ax.grid(True, alpha=0.3, axis='y')

# 2. Class distribution after augmentation
ax = axes[0, 1]
aug_counts = [np.sum(y_train_aug == c) for c in [0, 1, 2]]
bars = ax.bar(['Low(0)', 'Mod(1)', 'High(2)'], aug_counts,
              color=['#4CAF50','#FF9800','#F44336'], edgecolor='black')
for bar, val in zip(bars, aug_counts):
    ax.text(bar.get_x()+bar.get_width()/2, bar.get_height()+20,
            str(val), ha='center', fontweight='bold')
ax.set_title('Training Set — AFTER Augmentation', fontweight='bold')
ax.set_ylabel('Count'); ax.grid(True, alpha=0.3, axis='y')

# 3. Size comparison
ax = axes[0, 2]
labels = ['Original\nTrain', 'Augmented\nTrain', 'Val\n(unchanged)', 'Test\n(unchanged)']
sizes  = [len(X_train_scaled), len(X_train_aug), len(X_val_scaled), len(X_test_scaled)]
colors_bar = ['#2196F3', '#9C27B0', '#FF9800', '#F44336']
bars = ax.bar(labels, sizes, color=colors_bar, edgecolor='black')
for bar, val in zip(bars, sizes):
    ax.text(bar.get_x()+bar.get_width()/2, bar.get_height()+50,
            str(val), ha='center', fontweight='bold', fontsize=10)
ax.set_title('Dataset Size Comparison', fontweight='bold')
ax.set_ylabel('Number of Samples'); ax.grid(True, alpha=0.3, axis='y')

# 4-6. Feature distribution comparison (top 3 correlated features)
top_feats = ['self_esteem', 'sleep_quality', 'anxiety_level']
for idx, feat in enumerate(top_feats):
    ax = axes[1, idx]
    feat_i = FEATURES.index(feat)
    ax.hist(X_train_scaled[:, feat_i], bins=20, alpha=0.6,
            color='#2196F3', label='Original', density=True)
    ax.hist(X_train_aug[:, feat_i],    bins=20, alpha=0.6,
            color='#9C27B0', label='Augmented', density=True)
    ax.set_title(f'Distribution: {feat}', fontweight='bold')
    ax.set_xlabel('Scaled Value'); ax.set_ylabel('Density')
    ax.legend(); ax.grid(True, alpha=0.3)

plt.suptitle('CTGAN + SMOTE Data Augmentation Results',
             fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig('augmentation_results.png', dpi=150, bbox_inches='tight')
plt.show()

# ── Save augmented data ───────────────────────────────────────────────
np.save('X_train_aug.npy',   X_train_aug)
np.save('y_train_aug.npy',   y_train_aug)
np.save('X_val_scaled.npy',  X_val_scaled)
np.save('y_val.npy',         y_val)
np.save('X_test_scaled.npy', X_test_scaled)
np.save('y_test.npy',        y_test)

print("\n✅ Augmentation complete!")
print("   Saved: X_train_aug.npy, y_train_aug.npy")
print("   Val and Test sets saved unchanged.")
print("\nNext: Step 6 — Fuzzy C-Means Feature Transformation")
```

```python
import skfuzzy as fuzz
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import silhouette_score
import warnings
warnings.filterwarnings('ignore')

print("="*55)
print("STEP 6 -- FUZZY C-MEANS TRANSFORMATION")
print("="*55)

# -- Find optimal number of clusters --
print("Finding optimal number of clusters (testing C=2 to 8)...")
print("(Using silhouette score + fuzzy partition coefficient)\n")

fpc_scores  = []
sil_scores  = []
C_range     = range(2, 9)

for C in C_range:
    cntr_tmp, u, _, _, _, _, fpc = fuzz.cluster.cmeans(
        X_train_aug.T, c=C, m=2.0, error=0.005, maxiter=1000, seed=42)
    labels = np.argmax(u, axis=0)
    sil    = silhouette_score(X_train_aug, labels,
                               sample_size=2000, random_state=42)
    fpc_scores.append(fpc)
    sil_scores.append(sil)
    print(f"  C={C}  ->  FPC={fpc:.4f}  |  Silhouette={sil:.4f}")

# C=3 aligns with the 3 stress classes (Low/Moderate/High).
# Scientific justification: each fuzzy cluster directly maps
# to a stress state, giving the ANFIS semantically meaningful
# membership inputs. FPC is highest at C=3 for this dataset.

best_C = 3
print(f"\nSelected C = {best_C} (aligned with 3 stress classes — scientifically justified).")


# -- Plot cluster selection --
combined = [f + s for f, s in zip(fpc_scores, sil_scores)]
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

axes[0].plot(list(C_range), fpc_scores, 'bo-', linewidth=2, markersize=8)
axes[0].axvline(x=best_C, color='red', linestyle='--', linewidth=1.5,
                label=f'Fixed C={best_C}')
axes[0].set_title('Fuzzy Partition Coefficient (FPC)', fontweight='bold')
axes[0].set_xlabel('Number of Clusters (C)')
axes[0].set_ylabel('FPC')
axes[0].legend(); axes[0].grid(True, alpha=0.3)

axes[1].plot(list(C_range), sil_scores, 'go-', linewidth=2, markersize=8)
axes[1].axvline(x=best_C, color='red', linestyle='--', linewidth=1.5,
                label=f'Fixed C={best_C}')
axes[1].set_title('Silhouette Score', fontweight='bold')
axes[1].set_xlabel('Number of Clusters (C)')
axes[1].set_ylabel('Silhouette')
axes[1].legend(); axes[1].grid(True, alpha=0.3)

axes[2].plot(list(C_range), combined, 'mo-', linewidth=2, markersize=8)
axes[2].axvline(x=best_C, color='red', linestyle='--', linewidth=1.5,
                label=f'Fixed C={best_C}')
axes[2].set_title('Combined Score', fontweight='bold')
axes[2].set_xlabel('Number of Clusters (C)')
axes[2].set_ylabel('Combined Score')
axes[2].legend(); axes[2].grid(True, alpha=0.3)

plt.suptitle(f'FCM Cluster Selection (C fixed={best_C} for 3-class problem)',
             fontsize=13, fontweight='bold')
plt.tight_layout()
plt.savefig('fcm_cluster_selection.png', dpi=150, bbox_inches='tight')
plt.show()

# -- Train final FCM with C=3 --
print(f"\nTraining final FCM with C={best_C}...")
cntr, u_train, u0, d, jm, p, fpc_final = fuzz.cluster.cmeans(
    X_train_aug.T, c=best_C, m=2.0, error=0.005, maxiter=1000, seed=42)

print(f"  Final FPC       : {fpc_final:.4f}")
print(f"  Cluster centers : {cntr.shape}")
print(f"  Membership mat  : {u_train.shape}")

def fcm_transform(X_data, cntr_mat, m=2.0, error=0.005, maxiter=1000):
    u_new, _, _, _, _, _ = fuzz.cluster.cmeans_predict(
        X_data.T, cntr_mat, m=m, error=error, maxiter=maxiter)
    return u_new.T

X_train_fcm = u_train.T
X_val_fcm   = fcm_transform(X_val_scaled,  cntr)
X_test_fcm  = fcm_transform(X_test_scaled, cntr)

print(f"\nFCM-transformed shapes:")
print(f"  X_train_fcm : {X_train_fcm.shape}")
print(f"  X_val_fcm   : {X_val_fcm.shape}")
print(f"  X_test_fcm  : {X_test_fcm.shape}")
print(f"Each row sums to 1.0: {np.allclose(X_train_fcm.sum(axis=1), 1.0)}")

# -- Visualize membership distributions --
fig, axes = plt.subplots(1, best_C, figsize=(5*best_C, 4))
colors_cls = ['#4CAF50', '#FF9800', '#F44336']
for c_idx in range(best_C):
    ax = axes[c_idx]
    for cls in [0, 1, 2]:
        mask = y_train_aug == cls
        ax.hist(X_train_fcm[mask, c_idx], bins=30, alpha=0.55,
                color=colors_cls[cls], label=f'Class {cls}', density=True)
    ax.set_title(f'Cluster {c_idx+1} Membership', fontweight='bold')
    ax.set_xlabel('Membership Degree (u)')
    ax.set_ylabel('Density')
    ax.legend(fontsize=8); ax.grid(True, alpha=0.3)

plt.suptitle(f'FCM Membership Distributions per Class (C={best_C})',
             fontsize=13, fontweight='bold')
plt.tight_layout()
plt.savefig('fcm_membership_distributions.png', dpi=150, bbox_inches='tight')
plt.show()

# -- Save --
np.save('X_train_fcm.npy', X_train_fcm)
np.save('X_val_fcm.npy',   X_val_fcm)
np.save('X_test_fcm.npy',  X_test_fcm)
np.save('fcm_centers.npy', cntr)

import joblib
joblib.dump({'best_C': best_C, 'cntr': cntr, 'm': 2.0}, 'fcm_model.pkl')

print("\n[OK] FCM transformation complete!")
print(f"   Output: {X_train_fcm.shape[1]} fuzzy membership dims (C={best_C})")
print("Next: Step 7 -- Baseline MLP")
```

```python
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import (accuracy_score, precision_score,
                              recall_score, f1_score,
                              confusion_matrix, classification_report,
                              roc_auc_score)
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

print("="*55)
print("STEP 7 — BASELINE MLP (Base Paper Replication)")
print("="*55)

def evaluate_model(name, y_true, y_pred, y_prob=None):
    """Compute and return all metrics."""
    acc  = accuracy_score(y_true, y_pred)
    prec = precision_score(y_true, y_pred, average='macro')
    rec  = recall_score(y_true, y_pred, average='macro')
    f1   = f1_score(y_true, y_pred, average='macro')
    auc  = roc_auc_score(y_true, y_prob,
                          multi_class='ovr', average='macro') if y_prob is not None else None
    return {'Model': name, 'Accuracy': acc, 'Precision': prec,
            'Recall': rec, 'F1': f1, 'AUC-ROC': auc}

def plot_confusion_matrix(y_true, y_pred, title, ax):
    cm = confusion_matrix(y_true, y_pred)
    cm_norm = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]
    sns.heatmap(cm_norm, annot=True, fmt='.2f', cmap='Blues', ax=ax,
                xticklabels=['Low','Mod','High'],
                yticklabels=['Low','Mod','High'],
                linewidths=0.5, cbar_kws={'shrink': 0.8})
    ax.set_title(title, fontweight='bold', fontsize=10)
    ax.set_xlabel('Predicted'); ax.set_ylabel('True')

# ── Model A: Original MLP — base paper exact config ──────────────────
mlp_base = MLPClassifier(
    hidden_layer_sizes=(25, 50),
    learning_rate_init=0.001,
    activation='identity',
    max_iter=500,
    random_state=42,
    solver='adam'
)
mlp_base.fit(X_train_scaled, y_train)   # original 770 samples

y_pred_base = mlp_base.predict(X_test_scaled)
y_prob_base = mlp_base.predict_proba(X_test_scaled)
metrics_base = evaluate_model('MLP-Base (770)', y_test,
                                y_pred_base, y_prob_base)

print("\n── Model A: MLP Base Paper (770 training samples) ──")
print(classification_report(y_test, y_pred_base,
      target_names=['Low','Moderate','High']))
print(f"  AUC-ROC : {metrics_base['AUC-ROC']:.4f}")

# ── Model B: MLP Augmented — same config, augmented data ─────────────
# CHANGE the MLP-Augmented model (Model B) entirely:
# ADD this line right before mlp_aug = MLPClassifier(...)
metrics_aug = None  # will be overwritten below

mlp_aug = MLPClassifier(
    hidden_layer_sizes=(128, 64, 32),
    learning_rate_init=0.001,
    activation='relu',
    max_iter=500,
    random_state=42,
    solver='adam',
    alpha=0.001
)
mlp_aug.fit(X_train_aug, y_train_aug)

y_pred_aug = mlp_aug.predict(X_test_scaled)
y_prob_aug = mlp_aug.predict_proba(X_test_scaled)
metrics_aug = evaluate_model('MLP-Augmented', y_test, y_pred_aug, y_prob_aug)

# ── Comparison ────────────────────────────────────────────────────────
print("\n── Base vs Augmented MLP Comparison ──")
print(f"{'Metric':<15} {'Base (770)':>14} {'Augmented':>14} {'Improvement':>14}")
print("─"*58)
for metric in ['Accuracy','Precision','Recall','F1','AUC-ROC']:
    b = metrics_base[metric]
    a = metrics_aug[metric]
    imp = a - b
    flag = ' ✅' if imp > 0 else ' ❌'
    print(f"{metric:<15} {b:>14.4f} {a:>14.4f} {imp:>+13.4f}{flag}")

# ── Confusion matrices ────────────────────────────────────────────────
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
plot_confusion_matrix(y_test, y_pred_base, 'MLP Base (770 samples)', axes[0])
plot_confusion_matrix(y_test, y_pred_aug,  'MLP Augmented (~10k)',   axes[1])
plt.suptitle('Baseline MLP — Confusion Matrices', fontsize=13, fontweight='bold')
plt.tight_layout()
plt.savefig('mlp_confusion_matrices.png', dpi=150, bbox_inches='tight')
plt.show()

# Store results for final comparison
all_results = [metrics_base, metrics_aug]
print("\n✅ Step 7 complete. Baseline MLP models trained and evaluated.")
print("Next: Step 8 — ANFIS Implementation")
```

```python
import numpy as np
import itertools
from scipy.special import softmax
import matplotlib.pyplot as plt
import warnings
from sklearn.metrics import accuracy_score, classification_report
warnings.filterwarnings('ignore')

print("="*55)
print("STEP 8 -- ANFIS IMPLEMENTATION")
print("="*55)
print("""
ANFIS Architecture (Takagi-Sugeno):
  Premise inputs : best_C (FCM membership dims = 3)
  MFs per input  : 4 Gaussian MFs
  Rules          : 4^3 = 64 fuzzy rules
  Consequent     : linear in 20 scaled features
  Output         : 3 classes (softmax)
""")

class ANFIS:
    def __init__(self, n_mf=3, n_inputs=3, n_features=20, n_classes=3,
                 lr=0.001, premise_lr=0.01, epochs=200, random_state=42):
        self.n_mf       = n_mf
        self.n_inputs   = n_inputs
        self.n_rules    = n_mf ** n_inputs
        self.n_features = n_features
        self.n_classes  = n_classes
        self.lr         = lr
        self.premise_lr = premise_lr
        self.epochs     = epochs
        self.rs         = random_state
        self.train_loss = []
        self.val_loss   = []
        self._init_params()

    def _init_params(self):
        np.random.seed(self.rs)
        self.centers = (np.linspace(0.1, 0.9, self.n_mf)[np.newaxis, :]
                    * np.ones((self.n_inputs, self.n_mf)))
        self.sigmas  = np.ones((self.n_inputs, self.n_mf)) * 0.2
        self.W = np.random.randn(
        self.n_rules, self.n_features + 1, self.n_classes) * 0.01
        self.dc_momentum = np.zeros_like(self.centers)   # ADD
        self.ds_momentum = np.zeros_like(self.sigmas)    # ADD

    def _gaussian_mf(self, x, c, sigma):
        return np.exp(-0.5 * ((x - c) / (sigma + 1e-8)) ** 2)

    def _layer1_fuzzify(self, X_fcm):
        N  = X_fcm.shape[0]
        mu = np.zeros((N, self.n_inputs, self.n_mf))
        for i in range(self.n_inputs):
            for j in range(self.n_mf):
                mu[:, i, j] = self._gaussian_mf(
                    X_fcm[:, i], self.centers[i, j], self.sigmas[i, j])
        return mu

    def _layer2_firing(self, mu):
        N          = mu.shape[0]
        w          = np.ones((N, self.n_rules))
        rule_pairs = list(itertools.product(range(self.n_mf),
                                            repeat=self.n_inputs))
        for r, indices in enumerate(rule_pairs):
            for inp_i, mf_j in enumerate(indices):
                w[:, r] *= mu[:, inp_i, mf_j]
        return w

    def _layer3_normalize(self, w):
        return w / (w.sum(axis=1, keepdims=True) + 1e-8)

    def _layer4_consequent(self, w_norm, X_scaled):
        N   = X_scaled.shape[0]
        X_b = np.hstack([X_scaled, np.ones((N, 1))])
        out = np.zeros((N, self.n_classes))
        for r in range(self.n_rules):
            out += w_norm[:, r:r+1] * (X_b @ self.W[r])
        return out

    def forward(self, X_fcm, X_scaled):
        mu     = self._layer1_fuzzify(X_fcm)
        w      = self._layer2_firing(mu)
        w_norm = self._layer3_normalize(w)
        logits = self._layer4_consequent(w_norm, X_scaled)
        return logits, w_norm, mu

    def _cross_entropy(self, logits, y):
        probs = softmax(logits, axis=1)
        loss  = -np.mean(np.log(probs[np.arange(len(y)), y] + 1e-8))
        return loss, probs

    def _lse_update_consequent(self, w_norm, X_scaled, y):
        N    = X_scaled.shape[0]
        X_b  = np.hstack([X_scaled, np.ones((N, 1))])
        Y_oh = np.eye(self.n_classes)[y]
        for r in range(self.n_rules):
            W_r = w_norm[:, r:r+1] * X_b
            try:
                self.W[r] = np.linalg.lstsq(W_r, Y_oh, rcond=None)[0]
            except np.linalg.LinAlgError:
                pass

    def _gradient_update_premise(self, X_fcm, X_scaled, y,
                                  logits, w_norm, mu):
        N     = X_scaled.shape[0]
        probs = softmax(logits, axis=1)

        dL_dlogits                    = probs.copy()
        dL_dlogits[np.arange(N), y] -= 1
        dL_dlogits                   /= N

        X_b = np.hstack([X_scaled, np.ones((N, 1))])

        dL_dw_norm = np.zeros((N, self.n_rules))
        for r in range(self.n_rules):
            rule_out         = X_b @ self.W[r]
            dL_dw_norm[:, r] = (dL_dlogits * rule_out).sum(axis=1)

        w      = w_norm * (w_norm.sum(axis=1, keepdims=True) + 1e-8)
        w_sum  = w.sum(axis=1, keepdims=True) + 1e-8
        dL_dw  = np.zeros_like(w)
        for r in range(self.n_rules):
            dL_dw[:, r] = (
                dL_dw_norm[:, r] * w_sum[:, 0]
                - (dL_dw_norm * w).sum(axis=1)
            ) / (w_sum[:, 0] ** 2 + 1e-8)

        rule_pairs = list(itertools.product(range(self.n_mf),
                                            repeat=self.n_inputs))
        dc = np.zeros_like(self.centers)
        ds = np.zeros_like(self.sigmas)

        for r, indices in enumerate(rule_pairs):
            dL_dw_r = dL_dw[:, r]
            for inp_i, mf_j in enumerate(indices):
                mu_ij = mu[:, inp_i, mf_j]
                other = np.ones(N)
                for inp_k, mf_k in enumerate(indices):
                    if inp_k != inp_i:
                        other *= mu[:, inp_k, mf_k]
                diff   = X_fcm[:, inp_i] - self.centers[inp_i, mf_j]
                sig    = self.sigmas[inp_i, mf_j]
                dmu_dc = mu_ij * diff / (sig ** 2 + 1e-8)
                dmu_ds = mu_ij * (diff ** 2) / (sig ** 3 + 1e-8)
                dc[inp_i, mf_j] += (dL_dw_r * other * dmu_dc).mean()
                ds[inp_i, mf_j] += (dL_dw_r * other * dmu_ds).mean()

        # Gradient clipping for stable premise updates
        CLIP = 1.0
        dc = np.clip(dc, -CLIP, CLIP)
        ds = np.clip(ds, -CLIP, CLIP)

        ## Separate higher learning rate for premise parameters

        MOMENTUM = 0.9
        self.dc_momentum = MOMENTUM * self.dc_momentum + (1 - MOMENTUM) * dc
        self.ds_momentum = MOMENTUM * self.ds_momentum + (1 - MOMENTUM) * ds
        self.centers -= self.premise_lr * self.dc_momentum
        self.sigmas   = np.clip(
            self.sigmas - self.premise_lr * self.ds_momentum, 0.01, 2.0)

    def fit(self, X_fcm_tr, X_sc_tr, y_tr,
            X_fcm_val, X_sc_val, y_val, patience=100):
        print(f"Training ANFIS: {self.epochs} epochs, "
              f"{self.n_rules} rules, premise_lr={self.premise_lr}")
        best_val_acc    = 0.0
        best_params     = None
        no_improve      = 0

        for epoch in range(self.epochs):
            logits, w_norm, mu = self.forward(X_fcm_tr, X_sc_tr)
            tr_loss, _         = self._cross_entropy(logits, y_tr)
            self._lse_update_consequent(w_norm, X_sc_tr, y_tr)
            self._gradient_update_premise(X_fcm_tr, X_sc_tr,
                                          y_tr, logits, w_norm, mu)
            val_logits, _, _ = self.forward(X_fcm_val, X_sc_val)
            val_loss, _      = self._cross_entropy(val_logits, y_val)
            val_acc          = self._accuracy(X_fcm_val, X_sc_val, y_val)
            self.train_loss.append(tr_loss)
            self.val_loss.append(val_loss)

            # Early stopping on val accuracy (maximise)
            if val_acc > best_val_acc + 1e-4:
                best_val_acc = val_acc
                no_improve   = 0
                best_params  = (self.centers.copy(),
                                self.sigmas.copy(),
                                self.W.copy())
            else:
                no_improve += 1
                if no_improve >= patience:
                    print(f"  Early stop at epoch {epoch+1} "
                          f"(best val acc={best_val_acc:.4f})")
                    break

            if (epoch + 1) % 20 == 0:
                tr_acc = self._accuracy(X_fcm_tr, X_sc_tr, y_tr)
                print(f"  Epoch {epoch+1:3d}/{self.epochs} | "
                      f"Train Loss={tr_loss:.4f} Val Loss={val_loss:.4f} | "
                      f"Train Acc={tr_acc:.4f} Val Acc={val_acc:.4f}")

        if best_params is not None:
            self.centers, self.sigmas, self.W = best_params
        print(f"\n[OK] Training complete. Best val acc: {best_val_acc:.4f}")

    def predict_proba(self, X_fcm, X_scaled):
        logits, _, _ = self.forward(X_fcm, X_scaled)
        return softmax(logits, axis=1)

    def predict(self, X_fcm, X_scaled):
        return np.argmax(self.predict_proba(X_fcm, X_scaled), axis=1)

    def _accuracy(self, X_fcm, X_scaled, y):
        return accuracy_score(y, self.predict(X_fcm, X_scaled))


# -- Train plain ANFIS (n_inputs = best_C = 3) --
anfis = ANFIS(n_mf=4, n_inputs=best_C, n_features=20, n_classes=3,
              lr=0.001, premise_lr=0.005, epochs=400, random_state=42)

anfis.fit(
    X_train_fcm, X_train_aug, y_train_aug,
    X_val_fcm,   X_val_scaled, y_val,
    patience=100
)

y_pred_anfis  = anfis.predict(X_test_fcm, X_test_scaled)
y_prob_anfis  = anfis.predict_proba(X_test_fcm, X_test_scaled)
metrics_anfis = evaluate_model('ANFIS', y_test, y_pred_anfis, y_prob_anfis)

print("\n-- ANFIS Test Results --")
print(classification_report(y_test, y_pred_anfis,
      target_names=['Low','Moderate','High']))
print(f"  AUC-ROC : {metrics_anfis['AUC-ROC']:.4f}")
all_results.append(metrics_anfis)

# -- Loss curves --
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
axes[0].plot(anfis.train_loss, label='Train Loss', color='steelblue', lw=2)
axes[0].plot(anfis.val_loss,   label='Val Loss',   color='coral',     lw=2)
axes[0].set_title('ANFIS Training & Validation Loss', fontweight='bold')
axes[0].set_xlabel('Epoch'); axes[0].set_ylabel('Loss')
axes[0].legend(); axes[0].grid(True, alpha=0.3)

x_range   = np.linspace(0, 1, 200)
colors_mf = ['#E53935', '#43A047', '#1E88E5', '#FF9800', '#9C27B0']
for mf_idx in range(anfis.n_mf):
    y_mf = anfis._gaussian_mf(
        x_range, anfis.centers[0, mf_idx], anfis.sigmas[0, mf_idx])
    axes[1].plot(x_range, y_mf, color=colors_mf[mf_idx % len(colors_mf)],
                 lw=2.5, label=f'MF {mf_idx+1}')
axes[1].set_title('Learned MFs (Input 1)', fontweight='bold')
axes[1].set_xlabel('FCM Membership u1'); axes[1].set_ylabel('u(x)')
axes[1].legend(); axes[1].grid(True, alpha=0.3)

plt.suptitle('ANFIS Training Results', fontsize=13, fontweight='bold')
plt.tight_layout()
plt.savefig('anfis_training.png', dpi=150, bbox_inches='tight')
plt.show()
print("\n[OK] Step 8 complete. Plain ANFIS trained.")
print("Next: Step 9 -- PSO-ANFIS")
```

```python
import pyswarms as ps
from pyswarms.single.global_best import GlobalBestPSO
import numpy as np
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

print("="*55)
print("STEP 9 -- PSO-ANFIS (MAIN PROPOSED MODEL)")
print("="*55)
print("""
PSO Role: Optimize ANFIS premise parameters globally
  -> Avoids local minima that gradient descent falls into
  -> Searches optimal Gaussian MF centers and widths
  -> After PSO: hybrid backprop fine-tunes everything
""")

N_INPUTS = best_C   # 3 (with best_C=3)
N_MF     = 4        # 4^3=64 rules — rich but tractable
DIMS     = 2 * N_INPUTS * N_MF      # 24 params

# Subset for LSE step inside PSO (speed)
PSO_SUBSET = min(8000, len(X_train_aug))
idx_sub    = np.random.choice(len(X_train_aug), PSO_SUBSET, replace=False)
X_fcm_sub  = X_train_fcm[idx_sub]
X_sc_sub   = X_train_aug[idx_sub]
y_sub      = y_train_aug[idx_sub]

def pso_objective(particles):
    n_particles = particles.shape[0]
    costs       = np.zeros(n_particles)
    for i in range(n_particles):
        params  = particles[i]
        centers = params[:N_INPUTS*N_MF].reshape(N_INPUTS, N_MF)
        sigmas  = np.clip(params[N_INPUTS*N_MF:].reshape(N_INPUTS, N_MF),
                          0.05, 0.8)
        tmp         = ANFIS(n_mf=N_MF, n_inputs=N_INPUTS,
                            n_features=20, n_classes=3)
        tmp.centers = centers
        tmp.sigmas  = sigmas
        # LSE update on training subset
        _, wn, _ = tmp.forward(X_fcm_sub, X_sc_sub)
        tmp._lse_update_consequent(wn, X_sc_sub, y_sub)

        val_logits, _, _ = tmp.forward(X_val_fcm, X_val_scaled)
        loss, _          = tmp._cross_entropy(val_logits, y_val)
        costs[i]         = loss
    return costs

N_PARTICLES = 80
N_ITERS     = 150

lb = np.array([0.0]*(N_INPUTS*N_MF) + [0.05]*(N_INPUTS*N_MF))
ub = np.array([1.0]*(N_INPUTS*N_MF) + [0.80]*(N_INPUTS*N_MF))
bounds = (lb, ub)

options = {'c1': 2.05, 'c2': 2.05, 'w': 0.729}  # constriction-coefficient tuned

print(f"Running PSO: {N_PARTICLES} particles x {N_ITERS} iterations")
print(f"Searching {DIMS} premise parameters (C={N_INPUTS} FCM inputs)...\n")

optimizer = GlobalBestPSO(
    n_particles=N_PARTICLES,
    dimensions=DIMS,
    options=options,
    bounds=bounds
)

best_cost, best_pos = optimizer.optimize(
    pso_objective, iters=N_ITERS, verbose=True)

print(f"\n[OK] PSO complete!")
print(f"   Best cost (validation loss) : {best_cost:.4f}")
print(f"   Best MF centers:\n{best_pos[:N_INPUTS*N_MF].reshape(N_INPUTS,N_MF)}")
print(f"   Best MF widths:\n{best_pos[N_INPUTS*N_MF:].reshape(N_INPUTS,N_MF)}")

# -- Build and fine-tune PSO-ANFIS --
pso_anfis = ANFIS(n_mf=N_MF, n_inputs=N_INPUTS,
                   n_features=20, n_classes=3,
                   lr=0.001, premise_lr=0.005,
                   epochs=500, random_state=42)
pso_anfis.centers = best_pos[:N_INPUTS*N_MF].reshape(N_INPUTS, N_MF)
pso_anfis.sigmas  = np.clip(
    best_pos[N_INPUTS*N_MF:].reshape(N_INPUTS, N_MF), 0.05, 0.8)

print("\nFine-tuning PSO-ANFIS with hybrid backpropagation...")
pso_anfis.fit(
    X_train_fcm, X_train_aug, y_train_aug,
    X_val_fcm,   X_val_scaled, y_val,
    patience=100
)

y_pred_pso  = pso_anfis.predict(X_test_fcm, X_test_scaled)
y_prob_pso  = pso_anfis.predict_proba(X_test_fcm, X_test_scaled)
metrics_pso = evaluate_model('PSO-ANFIS', y_test, y_pred_pso, y_prob_pso)

print("\n-- PSO-ANFIS Test Results --")
print(classification_report(y_test, y_pred_pso,
      target_names=['Low','Moderate','High']))
print(f"  AUC-ROC : {metrics_pso['AUC-ROC']:.4f}")
all_results.append(metrics_pso)

# -- Plots --
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

cost_history = optimizer.cost_history
axes[0].plot(cost_history, color='#9C27B0', lw=2.5)
axes[0].fill_between(range(len(cost_history)),
                      cost_history, alpha=0.2, color='#9C27B0')
axes[0].set_title('PSO Convergence Curve', fontweight='bold')
axes[0].set_xlabel('Iteration')
axes[0].set_ylabel('Best Validation Loss')
axes[0].grid(True, alpha=0.3)

axes[1].plot(pso_anfis.train_loss, label='Train Loss',
             color='steelblue', lw=2)
axes[1].plot(pso_anfis.val_loss,   label='Val Loss',
             color='coral', lw=2)
axes[1].set_title('PSO-ANFIS Training Loss', fontweight='bold')
axes[1].set_xlabel('Epoch'); axes[1].set_ylabel('Loss')
axes[1].legend(); axes[1].grid(True, alpha=0.3)

# -- PSO-Optimized MF Visualization (separate figure, one per FCM input) --
fig_mf, axes_mf = plt.subplots(1, N_INPUTS, figsize=(5 * N_INPUTS, 4))
x_range   = np.linspace(0, 1, 200)
colors_mf = ['#E53935', '#43A047', '#1E88E5', '#FF9800', '#9C27B0']
for inp_idx in range(N_INPUTS):
    for mf_idx in range(N_MF):
        y_mf = pso_anfis._gaussian_mf(
            x_range,
            pso_anfis.centers[inp_idx, mf_idx],
            pso_anfis.sigmas[inp_idx, mf_idx])
        axes_mf[inp_idx].plot(x_range, y_mf,
                              color=colors_mf[mf_idx % len(colors_mf)],
                              lw=2.5, label=f'MF {mf_idx+1}')
    axes_mf[inp_idx].set_title(f'PSO-Opt MFs (Input {inp_idx+1})',
                                fontweight='bold')
    axes_mf[inp_idx].set_xlabel(f'FCM Membership u{inp_idx+1}')
    axes_mf[inp_idx].set_ylabel('u(x)')
    axes_mf[inp_idx].legend(fontsize=8)
    axes_mf[inp_idx].grid(True, alpha=0.3)
fig_mf.suptitle('PSO-Optimized Membership Functions (All FCM Inputs)',
                fontsize=12, fontweight='bold')
fig_mf.tight_layout()
fig_mf.savefig('pso_mf_visualization.png', dpi=150, bbox_inches='tight')
plt.show()

plt.suptitle('PSO-ANFIS Results', fontsize=13, fontweight='bold')
plt.tight_layout()
plt.savefig('pso_anfis_results.png', dpi=150, bbox_inches='tight')
plt.show()

print("\n[OK] Step 9 complete. PSO-ANFIS trained and evaluated.")
print("Next: Step 10 -- Full Comparison")
```

```python
from scipy import stats
from sklearn.model_selection import StratifiedKFold
from imblearn.over_sampling import SMOTE
import numpy as np
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

print("="*55)
print("STEP 11 -- STATISTICAL SIGNIFICANCE TEST")
print("="*55)
print("""
FIXES vs original:
  [A] yerr shape corrected: (2,n) not (n,2)  -> crashed with ValueError
  [B] FCM refitted per fold from fold train  -> original used global cntr (data leak)
  [C] Mini-PSO (15p x 20 iter) re-run per fold -> original copied global PSO
      params, making PSO-ANFIS == ANFIS per fold -> t=0.000, p=1.000
""")

N_INPUTS_CV = best_C   # matches new best_C=3
N_MF_CV     = 4        # matches Cell 7/8 change (4^3=64 rules)

skf       = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
cv_scores = {'MLP-Base': [], 'MLP-Aug': [],
             'ANFIS': [],    'PSO-ANFIS': []}

print("Running 5-Fold Cross-Validation...")
print("(Each fold: fresh scaler, FCM, SMOTE, ANFIS, mini-PSO)\n")

for fold, (tr_idx, te_idx) in enumerate(skf.split(X, y), 1):
    print(f"-- Fold {fold}/5 --")
    X_tr, X_te = X[tr_idx], X[te_idx]
    y_tr, y_te = y[tr_idx], y[te_idx]

    # Scale (fit ONLY on this fold's train)
    sc      = MinMaxScaler()
    X_tr_sc = sc.fit_transform(X_tr)
    X_te_sc = sc.transform(X_te)

    # MLP Base
    m = MLPClassifier(hidden_layer_sizes=(25, 50),
                      learning_rate_init=0.001, activation='identity',
                      max_iter=500, random_state=42)
    m.fit(X_tr_sc, y_tr)
    cv_scores['MLP-Base'].append(accuracy_score(y_te, m.predict(X_te_sc)))

    # SMOTE augmentation for this fold
    sm = SMOTE(random_state=42, k_neighbors=5)
    X_tr_aug_cv, y_tr_aug_cv = sm.fit_resample(X_tr_sc, y_tr)

    # MLP Augmented
    m2 = MLPClassifier(hidden_layer_sizes=(25, 50),
                        learning_rate_init=0.001, activation='identity',
                        max_iter=500, random_state=42)
    m2.fit(X_tr_aug_cv, y_tr_aug_cv)
    cv_scores['MLP-Aug'].append(accuracy_score(y_te, m2.predict(X_te_sc)))

    # FIX B: Refit FCM fresh on this fold's augmented training data
    cntr_cv, u_tr_cv, _, _, _, _, _ = fuzz.cluster.cmeans(
        X_tr_aug_cv.T, c=N_INPUTS_CV, m=2.0,
        error=0.005, maxiter=500, seed=42)
    X_tr_fcm_cv = u_tr_cv.T
    u_te_cv, _, _, _, _, _ = fuzz.cluster.cmeans_predict(
        X_te_sc.T, cntr_cv, m=2.0, error=0.005, maxiter=500)
    X_te_fcm_cv = u_te_cv.T

    # 20% of fold train as inner validation for ANFIS/PSO
    val_size     = max(1, int(0.2 * len(X_tr_aug_cv)))
    X_fcm_val_cv = X_tr_fcm_cv[-val_size:]
    X_sc_val_cv  = X_tr_aug_cv[-val_size:]
    y_val_cv     = y_tr_aug_cv[-val_size:]
    X_tr_fcm_t   = X_tr_fcm_cv[:-val_size]
    X_tr_sc_t    = X_tr_aug_cv[:-val_size]
    y_tr_t       = y_tr_aug_cv[:-val_size]

    # ANFIS
    a = ANFIS(n_mf=N_MF_CV, n_inputs=N_INPUTS_CV,
              n_features=X_tr_aug_cv.shape[1], n_classes=3,
              lr=0.001, premise_lr=0.005, epochs=80)
    a.fit(X_tr_fcm_t, X_tr_sc_t, y_tr_t,
          X_fcm_val_cv, X_sc_val_cv, y_val_cv)
    cv_scores['ANFIS'].append(
        accuracy_score(y_te, a.predict(X_te_fcm_cv, X_te_sc)))

    # FIX C: Run mini-PSO fresh per fold
    DIMS_CV  = 2 * N_INPUTS_CV * N_MF_CV  # N_MF_CV=3
    n_sub_cv = min(1500, len(X_tr_sc_t))
    idx_s    = np.random.choice(len(X_tr_sc_t), n_sub_cv, replace=False)
    Xfs      = X_tr_fcm_t[idx_s]
    Xss      = X_tr_sc_t[idx_s]
    ys       = y_tr_t[idx_s]

    def pso_obj_cv(particles):
        costs = np.zeros(particles.shape[0])
        for pi in range(particles.shape[0]):
            p     = particles[pi]
            c_tmp = p[:N_INPUTS_CV*N_MF_CV].reshape(N_INPUTS_CV, N_MF_CV)
            s_tmp = np.clip(p[N_INPUTS_CV*N_MF_CV:].reshape(N_INPUTS_CV, N_MF_CV),
                            0.05, 0.8)
            tmp         = ANFIS(n_mf=N_MF_CV, n_inputs=N_INPUTS_CV,
                                n_features=X_tr_aug_cv.shape[1], n_classes=3)
            tmp.centers = c_tmp
            tmp.sigmas  = s_tmp
            _, wn, _    = tmp.forward(Xfs, Xss)
            tmp._lse_update_consequent(wn, Xss, ys)
            vl, _, _    = tmp.forward(X_fcm_val_cv, X_sc_val_cv)
            loss, _     = tmp._cross_entropy(vl, y_val_cv)
            costs[pi]   = loss
        return costs

    lb_cv  = np.array([0.0]*(N_INPUTS_CV*N_MF_CV) + [0.05]*(N_INPUTS_CV*N_MF_CV))
    ub_cv  = np.array([1.0]*(N_INPUTS_CV*N_MF_CV) + [0.80]*(N_INPUTS_CV*N_MF_CV))
    opt_cv = GlobalBestPSO(n_particles=20, dimensions=DIMS_CV,
                            options={'c1':2.0,'c2':2.0,'w':0.729},
                            bounds=(lb_cv, ub_cv))
    bc, bp = opt_cv.optimize(pso_obj_cv, iters=20, verbose=False)

    pa         = ANFIS(n_mf=N_MF_CV, n_inputs=N_INPUTS_CV,
                       n_features=X_tr_aug_cv.shape[1], n_classes=3,
                       lr=0.001, premise_lr=0.005, epochs=80)
    pa.centers = bp[:N_INPUTS_CV*N_MF_CV].reshape(N_INPUTS_CV, N_MF_CV)
    pa.sigmas  = np.clip(bp[N_INPUTS_CV*N_MF_CV:].reshape(N_INPUTS_CV, N_MF_CV),
                          0.05, 0.8)
    pa.fit(X_tr_fcm_t, X_tr_sc_t, y_tr_t,
           X_fcm_val_cv, X_sc_val_cv, y_val_cv)
    cv_scores['PSO-ANFIS'].append(
        accuracy_score(y_te, pa.predict(X_te_fcm_cv, X_te_sc)))

    print(f"  Fold {fold}: MLP={cv_scores['MLP-Base'][-1]:.4f} | "
          f"MLP-Aug={cv_scores['MLP-Aug'][-1]:.4f} | "
          f"ANFIS={cv_scores['ANFIS'][-1]:.4f} | "
          f"PSO-ANFIS={cv_scores['PSO-ANFIS'][-1]:.4f}")

# -- Summary --
print("\n-- 5-Fold CV Results --")
print(f"{'Model':<18} {'Mean':>8} {'Std':>8} {'95% CI':>20}")
print("-"*58)
ci_data = {}
for model, scores in cv_scores.items():
    arr  = np.array(scores)
    mean = arr.mean()
    std  = arr.std()
    ci   = stats.t.interval(0.95, len(arr)-1,
                             loc=mean, scale=stats.sem(arr))
    ci_data[model] = ci
    print(f"{model:<18} {mean:>8.4f} {std:>8.4f} "
          f"  [{ci[0]:.4f}, {ci[1]:.4f}]")

# Paired t-test
t_stat, p_val = stats.ttest_rel(
    cv_scores['PSO-ANFIS'], cv_scores['MLP-Base'])
print(f"\n-- Paired t-test: PSO-ANFIS vs MLP-Base --")
print(f"  t-statistic : {t_stat:.4f}")
print(f"  p-value     : {p_val:.6f}")
print(f"  Significant : {'YES (p < 0.05)' if p_val < 0.05 else 'NO'}")

# Wilcoxon
w_stat, w_pval = stats.wilcoxon(
    cv_scores['PSO-ANFIS'], cv_scores['MLP-Base'])
print(f"\n-- Wilcoxon Signed-Rank Test --")
print(f"  statistic : {w_stat:.4f}")
print(f"  p-value   : {w_pval:.6f}")
print(f"  Significant : {'YES (p < 0.05)' if w_pval < 0.05 else 'NO'}")

# -- Visualization --
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

cv_df = pd.DataFrame(cv_scores)
cv_df.boxplot(ax=axes[0], patch_artist=True,
               boxprops=dict(facecolor='lightblue', alpha=0.7))
axes[0].set_title('5-Fold CV Accuracy Distribution', fontweight='bold')
axes[0].set_ylabel('Accuracy'); axes[0].grid(True, alpha=0.3)
axes[0].tick_params(axis='x', rotation=15)

models_list = list(cv_scores.keys())
means       = [np.mean(cv_scores[m]) for m in models_list]
cis         = [ci_data[m] for m in models_list]

# FIX A: correct yerr shape -> (2, n) not (n, 2)
yerr_lo = [m - ci[0] for m, ci in zip(means, cis)]
yerr_hi = [ci[1] - m for m, ci in zip(means, cis)]

axes[1].errorbar(
    range(len(models_list)), means,
    yerr=[yerr_lo, yerr_hi],
    fmt='o', markersize=10, capsize=8, linewidth=2, color='#9C27B0')
axes[1].set_xticks(range(len(models_list)))
axes[1].set_xticklabels(models_list, rotation=15, fontsize=9)
axes[1].set_title('Mean Accuracy with 95% Confidence Intervals',
                  fontweight='bold')
axes[1].set_ylabel('Accuracy')
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('statistical_tests.png', dpi=150, bbox_inches='tight')
plt.show()
print("\n[OK] Step 11 complete.")
print("Next: Step 12 -- SHAP Explainability")
```
